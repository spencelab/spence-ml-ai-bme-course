# 3. Zero-based indexing and slicing

Python counts from **0**.

For:

```python
x = [10, 20, 30, 40, 50]
```

the positions are:

| value | 10 | 20 | 30 | 40 | 50 |
|---|---:|---:|---:|---:|---:|
| Python index | 0 | 1 | 2 | 3 | 4 |

Negative indices count backward from the end.

````python
x = [10, 20, 30, 40, 50]

print("first:", x[0])
print("third:", x[2])
print("last:", x[-1])

# slicing: endpoint is NOT included
print("indices 1,2,3:", x[1:4])
````

## MATLAB → NumPy indexing

If you know MATLAB, these translations are worth memorizing:

| Goal | MATLAB | NumPy |
|---|---|---|
| first element | `A(1)` | `A[0]` |
| row 1, col 1 | `A(1,1)` | `A[0,0]` |
| first column | `A(:,1)` | `A[:,0]` |
| first row | `A(1,:)` | `A[0,:]` |
| rows 2 through end | `A(2:end,:)` | `A[1:,:]` |
| columns 2–3 | `A(:,2:3)` | `A[:,1:3]` |

Notice two important differences:

1. Python starts counting at **0**
2. Python slices exclude the ending index

````python
A = np.array([
    [10, 11, 12],
    [20, 21, 22],
    [30, 31, 32],
    [40, 41, 42],
])

print("A:")
print(A)

print("\nFirst column:")
print(A[:, 0])

print("\nLast row:")
print(A[-1, :])

print("\nRows 2 through end:")
print(A[1:, :])
````

# 4. Same-looking data, different behavior

Here is a subtle but important difference.

Try comparing a Python list directly with a number:

````python
# Uncomment this line and run it:
# speeds_list > 1.0
````

A regular Python list does not know how to apply `> 1.0` element-by-element.

A NumPy array does:

````python
speeds_array > 1.0
````

The result is another array:

```text
[False  True  True  False]
```

That is called a **Boolean mask**.

It answers the question:

> For each observation, is this condition true or false?

````python
mask = speeds_array > 1.0

print("mask:")
print(mask)

print("\nvalues where mask is True:")
print(speeds_array[mask])
````

# 5. Boolean filtering in pandas

The same basic idea powers one of the most common operations in data science.

First, ask a question about one column:

````python
demo_df["walking_speed_m_s"] > 1.0
````

That produces a Boolean `Series`, one `True` or `False` for every row.

Use that Boolean Series inside square brackets to keep the matching rows:

````python
fast_trials = demo_df[demo_df["walking_speed_m_s"] > 1.0]
fast_trials
````

You can combine conditions.

For NumPy/pandas array conditions, use:

- `&` for AND
- `|` for OR
- `~` for NOT

and put each comparison in parentheses.

````python
demo_df[
    (demo_df["walking_speed_m_s"] > 0.85)
    & (demo_df["group"] == "SCI")
]
````

````python
print("First two rows by position:")
display(demo_df.iloc[0:2])

print("SCI rows using a Boolean condition:")
display(demo_df.loc[demo_df["group"] == "SCI"])
````

# 🚀 Stretch lane: use the same gait dataset for linear regression

Yes: the advanced students can use the **same synthetic gait dataset used in the boot camp**.

That is intentional. They can jump ahead into ML while everyone else works on the foundations, without changing datasets or context.

The next cell recreates the boot-camp dataset so this notebook can also work by itself.

````python
rng = np.random.default_rng(7)
n = 36

group = np.repeat(["Control", "SCI", "Assistive"], n // 3)

body_mass = rng.normal(72, 11, n)

days_post = np.concatenate([
    np.zeros(n // 3),
    rng.choice([7, 14, 21, 28, 42], n // 3),
    rng.choice([7, 14, 21, 28, 42], n // 3),
])

base_speed = np.where(
    group == "Control", 1.25,
    np.where(group == "SCI", 0.78, 0.98)
)

walking_speed = (
    base_speed
    + 0.006 * days_post
    + rng.normal(0, 0.10, n)
)

stride_length = (
    0.55
    + 0.58 * walking_speed
    + rng.normal(0, 0.06, n)
)

cadence = (
    72
    + 31 * walking_speed
    + rng.normal(0, 5, n)
)

stance_time = (
    0.92
    - 0.30 * walking_speed
    + rng.normal(0, 0.04, n)
)

gait = pd.DataFrame({
    "subject": [f"S{i+1:02d}" for i in range(n)],
    "group": group,
    "days_post_injury": days_post,
    "body_mass_kg": body_mass.round(1),
    "walking_speed_m_s": walking_speed.round(3),
    "stride_length_m": stride_length.round(3),
    "cadence_steps_min": cadence.round(1),
    "stance_time_s": stance_time.round(3),
})

# Add a few missing values, like real data
for row in [5, 17, 29]:
    gait.loc[row, "stride_length_m"] = np.nan

for row in [11, 24]:
    gait.loc[row, "cadence_steps_min"] = np.nan

gait.head()
````

## Stretch 2: linear regression

Use these predictors:

- `body_mass_kg`
- `stride_length_m`
- `cadence_steps_min`
- `stance_time_s`

to predict:

- `walking_speed_m_s`

Use a train/test split, median imputation, scaling, and linear regression.

If you already know scikit-learn, try doing this without looking below.

If you want a scaffold, expand/refer to the boot-camp example.

````python
# STRETCH

from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

features = [
    "body_mass_kg",
    "stride_length_m",
    "cadence_steps_min",
    "stance_time_s",
]

X = gait[features]
y = gait["walking_speed_m_s"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42
)

model = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
    LinearRegression(),
)

model.fit(X_train, y_train)

pred = model.predict(X_test)

rmse = mean_squared_error(y_test, pred) ** 0.5
r2 = r2_score(y_test, pred)

print("RMSE:", rmse)
print("R²:", r2)
````

When something fails, ask:

1. **What object do I actually have?**
   - list?
   - NumPy array?
   - Series?
   - DataFrame?

2. **What is its shape?**

3. **How is it indexed?**

4. **Am I asking for one value, a slice, a column, or rows matching a condition?**

Those four questions solve a shocking fraction of early scientific-Python bugs.
