- MATLAB `A(1,1)` → Python `A[0,0]`
- MATLAB `A(:,1)` → Python `A[:,0]`
- MATLAB `A(2:end,:)` → Python `A[1:,:]`

Python's ending slice index is **not included**.

### Element-wise math vs matrix multiplication
In NumPy, `A * B` is **element-wise multiplication** and `A @ B` is **matrix multiplication**. This differs from MATLAB.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("element-wise:")
print(A * B)

print("matrix multiply:")
print(A @ B)
```

# C. pandas + plotting: the everyday ML table
Most introductory ML datasets become a table where rows are observations and columns are variables/features. pandas gives us a `DataFrame` for this job.

## C0. Create a small biomechanics CSV for this boot camp
Run this setup cell. It creates a reproducible **synthetic** gait dataset and writes it to `gait_bootcamp.csv`. The point is that the next cell loads a CSV exactly the way we'll load real datasets.

```python
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
n = 36
group = np.repeat(["Control", "SCI", "Assistive"], n // 3)
body_mass = rng.normal(72, 11, n)
days_post = np.concatenate([
    np.zeros(n//3),
    rng.choice([7,14,21,28,42], n//3),
    rng.choice([7,14,21,28,42], n//3),
])
base_speed = np.where(group == "Control", 1.25, np.where(group == "SCI", 0.78, 0.98))
walking_speed = base_speed + 0.006*days_post + rng.normal(0,0.10,n)
stride_length = 0.55 + 0.58*walking_speed + rng.normal(0,0.06,n)
cadence = 72 + 31*walking_speed + rng.normal(0,5,n)
stance_time = 0.92 - 0.30*walking_speed + rng.normal(0,0.04,n)

df_setup = pd.DataFrame({
    "subject": [f"S{i+1:02d}" for i in range(n)],
    "group": group,
    "days_post_injury": days_post,
    "body_mass_kg": body_mass.round(1),
    "walking_speed_m_s": walking_speed.round(3),
    "stride_length_m": stride_length.round(3),
    "cadence_steps_min": cadence.round(1),
    "stance_time_s": stance_time.round(3),
})
for row in [5,17,29]: df_setup.loc[row,"stride_length_m"] = np.nan
for row in [11,24]: df_setup.loc[row,"cadence_steps_min"] = np.nan

df_setup.to_csv("gait_bootcamp.csv", index=False)
print("Created gait_bootcamp.csv")
```

# 🚀 Stretch C: jump ahead into ML
We'll predict walking speed from body mass, stride length, cadence, and stance time. This intentionally previews concepts we'll study properly later.

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

features = ["body_mass_kg", "stride_length_m", "cadence_steps_min", "stance_time_s"]
X = df[features]
y = df["walking_speed_m_s"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
model = make_pipeline(SimpleImputer(strategy="median"), StandardScaler(), LinearRegression())
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("RMSE:", mean_squared_error(y_test, pred) ** 0.5)
print("R²:", r2_score(y_test, pred))
```
